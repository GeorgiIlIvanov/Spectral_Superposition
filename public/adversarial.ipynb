```python
import numpy as np
import matplotlib.pyplot as plt

# Define the 7 classes (based on colors in your image)
categories = ['1', '2', '3', '4', '5', '6', '7']
N = len(categories)

# Data values
# Original: A single spike at Orange (0.5)
original_values = [0, 0.5, 0, 0, 0, 0, 0]

# Adversarial: Values provided (using absolute magnitudes for green/red shifts)
# Note: Using 0.06 and 0.05 for green/red as they represent the new presence in those classes
adversarial_values = [0.05, 0.47, 0.06, 0.05, 0.01, 0.02, 0.07]

# Repeat the first value to close the circular polygon
original_values += original_values[:1]
adversarial_values += adversarial_values[:1]

# Calculate angles for each category
angles = [n / float(N) * 2 * np.pi for n in range(N)]
angles += angles[:1]

# Plot setup
fig, ax = plt.subplots(figsize=(8, 8), subplot_kw=dict(polar=True))


ax.plot(angles, adversarial_values, color='purple', linewidth=2, label='Adversarial Input')
ax.fill(angles, adversarial_values, color='purple', alpha=0.25)

ax.plot(angles, original_values, color='orange', linewidth=2, label='Original Input')
ax.fill(angles, original_values, color='orange', alpha=0.25)

ax.set_xticks(angles[:-1])
ax.set_xticklabels(categories)
ax.set_ylim(0, 0.6)

plt.title("Adversarial Attack: Class Distribution Shift", size=15, y=1.1)
plt.legend(loc='upper right', bbox_to_anchor=(1.3, 1.1))

plt.show()
```
